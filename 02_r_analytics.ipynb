# Notebook 02 — R Analytics: Statistical Modelling and Visualisation

NorthStar Analytics — Database and Analytics Coursework 1

ANOVA, correlation analysis, multiple linear regression, and ggplot2 visualisations to quantify the patterns identified in Notebook 01.

```r
install.packages(c('tidyverse', 'broom', 'corrplot'), quiet = TRUE)
library(tidyverse); library(broom); library(corrplot)
set.seed(42)
```

```r
# Load and clean
customers  <- read.csv('customers.csv')
orders     <- read.csv('orders.csv')
deliveries <- read.csv('deliveries.csv')
drivers    <- read.csv('drivers.csv')
vehicles   <- read.csv('vehicles.csv')
hubs       <- read.csv('hubs.csv')

normalise_zone <- function(x) {
  x <- toupper(trimws(x))
  x[x %in% c('CTR','CENTRAL')] <- 'Central'
  x[x == 'AIRPORT']            <- 'Airport'
  x[x == 'NORTH']              <- 'North'
  x[x == 'SOUTH']              <- 'South'
  x[x == 'EAST']               <- 'East'
  x[x == 'WEST']               <- 'West'
  x[x == 'RIVERSIDE']          <- 'Riverside'
  return(x)
}
orders$pickup_zone <- normalise_zone(orders$pickup_zone)

master <- orders %>%
  inner_join(deliveries, by = 'order_id') %>%
  left_join(drivers,    by = 'driver_id') %>%
  left_join(vehicles,   by = 'vehicle_id') %>%
  left_join(hubs,       by = 'hub_id')

glimpse(master)
```

## 1. Descriptive statistics

```r
master %>%
  select(order_value, route_distance_km, fuel_or_charge_cost,
         manual_route_override_count, customer_rating_post_delivery) %>%
  summary()
```

## 2. ANOVA — Does delivery status affect customer rating?

```r
rating_aov <- aov(customer_rating_post_delivery ~ delivery_status, data = master)
summary(rating_aov)
TukeyHSD(rating_aov)
```

**Result:** F(2, 933) = 312.4, p < 2.2e-16. Mean ratings: OnTime 4.28, Delayed 3.12, Failed 3.05.

## 3. Correlation matrix

```r
install.packages('reshape2')
library(reshape2)
library(tidyverse)

perf_vars <- master %>%
  select(driver_rating, training_score, customer_rating_post_delivery,
         manual_route_override_count, route_distance_km,
         battery_health_pct, fuel_or_charge_cost) %>%
  na.omit()

cor_matrix <- cor(perf_vars, method = 'pearson')

colnames(cor_matrix) <- rownames(cor_matrix) <- c(
  'Driver Rating', 'Training Score', 'Customer Rating',
  'Overrides', 'Route Dist', 'Battery Health', 'Fuel Cost')

cor_melted <- melt(cor_matrix) %>%
  filter(Var1 != Var2)

ggplot(cor_melted, aes(x = Var1, y = Var2)) +
  geom_point(aes(size = abs(value), fill = value),
             shape = 21, colour = 'white', stroke = 0.5) +
  geom_text(aes(label = round(value, 2)),
            size = 2.8, colour = 'white', fontface = 'bold') +
  scale_fill_gradient2(low    = '#C1292E',
                        mid    = '#F5F5F5',
                        high   = '#003087',
                        midpoint = 0,
                        limits = c(-1, 1),
                        name   = 'Pearson r') +
  scale_size(range = c(3, 14), guide = 'none') +
  labs(title    = 'Figure 4.4 — Bubble Correlation Chart of NorthStar Operational Metrics',
       subtitle = 'Bubble size = strength of relationship | Blue = positive | Red = negative',
       caption  = 'Rahman Taiwo | Databases and Analytics CW1 | Source: NorthStar Dataset',
       x = '', y = '') +
  theme_minimal() +
  theme(
    axis.text.x  = element_text(angle = 30, hjust = 1, size = 9),
    axis.text.y  = element_text(size = 9),
    plot.title   = element_text(face = 'bold', size = 11),
    plot.subtitle = element_text(size = 9, colour = '#555555'),
    plot.caption = element_text(size = 8, colour = '#888888'),
    panel.grid   = element_blank()
  )
```

## 4. Multiple linear regression

```r
model <- lm(customer_rating_post_delivery ~
              driver_rating + training_score +
              manual_route_override_count + route_distance_km +
              battery_health_pct + fuel_or_charge_cost,
            data = master)

summary(model)
tidy(model)
glance(model)
```

```r
install.packages('ggfortify')
library(ggfortify)

autoplot(model,
         which  = 1:4,
         colour = '#0070C0',
         smooth.colour = '#C1292E',
         smooth.linetype = 'dashed',
         alpha  = 0.4) +
  theme_minimal() +
  labs(caption = 'Rahman Taiwo | Databases and Analytics CW1') +
  theme(plot.caption = element_text(size = 8, colour = '#888888'))
```

**Model summary:** R² = 0.301, Adj R² = 0.294, F(6, 922) = 66.21, p < 2.2e-16.

**Significant predictors:**
- driver_rating: β = 0.510, p < 0.001 (dominant)
- manual_overrides: β = -0.183, p = 0.003

**Not significant:** training_score, route_distance_km, fuel_or_charge_cost. The platform-rating mechanism is what moves customer satisfaction; classroom training expenditure does not.

## 5. ggplot2 visualisations

```r
# Figure 4.1 — Lollipop Chart: Failure rate by zone
zone_perf <- master %>%
  group_by(pickup_zone) %>%
  summarise(failure_rate_pct = round(100 * mean(delivery_status == 'Failed'), 2),
            n = n()) %>%
  arrange(desc(failure_rate_pct))

ggplot(zone_perf, aes(x = reorder(pickup_zone, failure_rate_pct),
                       y = failure_rate_pct)) +
  geom_segment(aes(xend = reorder(pickup_zone, failure_rate_pct),
                   y = 0, yend = failure_rate_pct,
                   colour = failure_rate_pct),
               size = 1.8) +
  geom_point(aes(colour = failure_rate_pct, size = n)) +
  geom_text(aes(label = paste0(failure_rate_pct, '%')),
            hjust = -0.3, fontface = 'bold', size = 3.5) +
  scale_colour_gradient(low = '#2A7F3A', high = '#C1292E',
                         name = 'Failure %') +
  scale_size(range = c(4, 10), name = 'Deliveries') +
  coord_flip() +
  ylim(0, 25) +
  labs(title    = 'Figure 4.1 — Delivery Failure Rate by Zone',
       subtitle = 'Dot size = delivery volume | Colour = severity\nCentral zone has the highest failure rate at 18.97%',
       x = '', y = 'Failure rate (%)',
       caption  = 'Rahman Taiwo | Databases and Analytics CW1 | Source: NorthStar Dataset') +
  theme_minimal() +
  theme(
    plot.title    = element_text(face = 'bold', size = 11),
    plot.subtitle = element_text(size = 9, colour = '#555555'),
    plot.caption  = element_text(size = 8, colour = '#888888'),
    axis.text     = element_text(size = 10),
    legend.position = 'right',
    panel.grid.minor = element_blank()
  )
```

```r
# Figure 4.2 — Faceted Scatter Plot: Driver rating vs Customer rating
# Split by delivery status to show relationship across all three outcomes

ggplot(master %>% filter(!is.na(customer_rating_post_delivery)),
       aes(x = driver_rating, y = customer_rating_post_delivery,
           colour = delivery_status)) +
  geom_point(alpha = 0.4, size = 1.8) +
  geom_smooth(method = 'lm', se = TRUE,
              colour = '#1A1A2E', fill = '#BBBBDD') +
  facet_wrap(~ delivery_status, ncol = 3) +
  scale_colour_manual(values = c('OnTime'  = '#2A7F3A',
                                  'Delayed' = '#F6AE2D',
                                  'Failed'  = '#C1292E'),
                       guide = 'none') +
  labs(title    = 'Figure 4.2 — Driver Platform Rating vs Post-Delivery Customer Rating',
       subtitle = 'Pearson r = 0.115, p < 0.001 | Faceted by delivery status\nRelationship holds across OnTime, Delayed and Failed deliveries',
       x        = 'Driver platform rating',
       y        = 'Post-delivery customer rating',
       caption  = 'Rahman Taiwo | Databases and Analytics CW1 | Source: NorthStar Dataset') +
  theme_minimal() +
  theme(
    plot.title    = element_text(face = 'bold', size = 11),
    plot.subtitle = element_text(size = 9, colour = '#555555'),
    plot.caption  = element_text(size = 8, colour = '#888888'),
    strip.text    = element_text(face = 'bold', size = 10),
    axis.text     = element_text(size = 9),
    panel.spacing = unit(1, 'lines')
  )
```

```r
# Failure rate by vehicle Maintenance Status
maint_summary <- master %>%
  group_by(maintenance_status) %>%
  summarise(failure_pct = round(100 * mean(delivery_status == 'Failed'), 2))

ggplot(maint_summary, aes(x = maintenance_status, y = failure_pct,
                          fill = maintenance_status)) +
  geom_col(width = 0.55) +
  geom_text(aes(label = paste0(failure_pct, '%')), vjust = -0.4, size = 4) +
  scale_fill_manual(values = c('Active'='#2A7F3A','Scheduled'='#0070C0','InRepair'='#C1292E')) +
  labs(title = 'Failure rate by vehicle maintenance status',
       subtitle = 'InRepair vehicles fail 3.6x more often than Active vehicles',
       x = 'Maintenance status', y = 'Failure rate (%)') +
  theme_minimal() + theme(legend.position = 'none')
```
